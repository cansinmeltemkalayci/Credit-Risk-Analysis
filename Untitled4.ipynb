# VAR Analysis: Monetary Policy & Macroeconomic Variables
# Data: FEDFUNDS, CPIAUCSL, INDPRO, DRCCLACBS (from FRED)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
from statsmodels.tsa.stattools import adfuller
from statsmodels.tsa.api import VAR
import warnings
warnings.filterwarnings("ignore")
```

#STEP 1: Load CSV files downloaded from FRED

```python
def load_fred_csv(filepath, colname):
    """
    Load a FRED CSV file.
    Reads the first column as date index regardless of its name,
    and renames the value column to colname.
    """
    df = pd.read_csv(filepath, na_values=".")
    # First column is always the date in FRED downloads
    date_col = df.columns[0]
    df[date_col] = pd.to_datetime(df[date_col])
    df = df.set_index(date_col)
    df.index.name = "DATE"
    # Keep only the first value column (second column in CSV)
    df = df.iloc[:, [0]]
    df.columns = [colname]
    return df
fedfunds = load_fred_csv("FEDFUNDS.csv",   "FEDFUNDS")
cpi      = load_fred_csv("CPIAUCSL.csv",   "CPIAUCSL")
indpro   = load_fred_csv("INDPRO.csv",     "INDPRO")
drc      = load_fred_csv("DRCCLACBS.csv",  "DRCCLACBS")
```

# STEP 2: Merge, resample to quarterly, drop NaNs

```python
data = fedfunds.join([cpi, indpro, drc], how="outer")
data = data.resample("QE").last()
data = data.dropna()
print(f"Dataset shape: {data.shape}")
print(f"Date range: {data.index[0].date()} to {data.index[-1].date()}")
print(data.tail())
```

# STEP 3: Apply transformations

```python
data["log_CPI"]    = 100 * np.log(data["CPIAUCSL"])
data["log_INDPRO"] = 100 * np.log(data["INDPRO"])
# Variable order for Cholesky: most exogenous first
df = data[["FEDFUNDS", "log_CPI", "log_INDPRO", "DRCCLACBS"]].copy()
print("\nTransformed data (last 5 rows):")
print(df.tail())
```

# STEP 4: ADF Test — first-difference non-stationary variables

```python
def adf_test(series, name, signif=0.05):
    """Run ADF test and return True if stationary at given significance level."""
    result = adfuller(series.dropna(), autolag="AIC")
    p_value = result[1]
    stationary = p_value < signif
    print(f"  {name}: ADF stat={result[0]:.4f}, p={p_value:.4f} "
          f"-> {'STATIONARY' if stationary else 'NON-STATIONARY'}")
    return stationary
print("\n--- ADF Tests (levels) ---")
stationary_flags = {}
for col in df.columns:
    stationary_flags[col] = adf_test(df[col], col)
df_model = df.copy()
differenced_vars = []
for col, is_stationary in stationary_flags.items():
    if not is_stationary:
        df_model[col] = df_model[col].diff()
        differenced_vars.append(col)
        print(f"  -> {col} first-differenced")
df_model = df_model.dropna()
if differenced_vars:
    print(f"\nVariables differenced: {differenced_vars}")
    print("--- ADF Tests (after differencing) ---")
    for col in differenced_vars:
        adf_test(df_model[col], f"D_{col}")
else:
    print("\nAll variables stationary in levels.")
print(f"\nFinal model dataset shape: {df_model.shape}")
```

# STEP 5: VAR Model — select optimal lag via AIC

```python
model = VAR(df_model)
lag_order_results = model.select_order(maxlags=8)
print("\n--- Lag Order Selection ---")
print(lag_order_results.summary())
optimal_lag = lag_order_results.aic
print(f"\nOptimal lag (AIC): {optimal_lag}")
```

# STEP 6: Estimate VAR model

```python
var_result = model.fit(maxlags=optimal_lag, ic="aic")
print(var_result.summary())
```

# STEP 7: Orthogonalized IRF (Cholesky) — FEDFUNDS as impulse

```python
PERIODS = 20
irf = var_result.irf(periods=PERIODS)
col_names = df_model.columns.tolist()
impulse_idx = col_names.index("FEDFUNDS")
var_labels = {
    "FEDFUNDS":   "Federal Funds Rate",
    "log_CPI":    "Log Price Level  100·ln(CPI)",
    "log_INDPRO": "Log Industrial Production  100·ln(INDPRO)",
    "DRCCLACBS":  "Credit Delinquency Rate",
}
n_vars = len(col_names)
fig, axes = plt.subplots(n_vars, 1, figsize=(10, 4 * n_vars), sharex=True)
fig.suptitle(
    "Orthogonalized IRF — Impulse: Federal Funds Rate (Cholesky)",
    fontsize=14, fontweight="bold"
)
periods_x = np.arange(PERIODS + 1)
for i, response_var in enumerate(col_names):
    ax = axes[i]
    irf_vals = irf.orth_irfs[:, i, impulse_idx]
    ax.plot(periods_x, irf_vals, color="steelblue", linewidth=2)
    ax.axhline(0, color="black", linewidth=0.8, linestyle="--")
    ax.fill_between(periods_x, irf_vals, 0,
                    where=(irf_vals > 0), alpha=0.15, color="steelblue")
    ax.fill_between(periods_x, irf_vals, 0,
                    where=(irf_vals < 0), alpha=0.15, color="tomato")
    label = var_labels.get(response_var, response_var)
    ax.set_title(f"Response of {label}", fontsize=11)
    ax.set_ylabel("Response", fontsize=9)
    ax.xaxis.set_major_locator(mticker.MultipleLocator(4))
    ax.grid(axis="x", linestyle=":", alpha=0.5)
axes[-1].set_xlabel("Quarters after shock", fontsize=11)
plt.tight_layout()
plt.savefig("irf_result.png", dpi=150, bbox_inches="tight")
plt.show()
print("\nSaved: irf_result.png")
```
